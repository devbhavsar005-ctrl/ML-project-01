```python
gender=df["gender"].unique()
print(f"unique value in gender :{gender}")

race_ethnicity=df["race_ethnicity"].unique()
print(f" unique in race_ethnicity:{race_ethnicity}")

parental_level_of_education=df["parental_level_of_education"].unique()
print(f"unique value in parental_level_of_education: {parental_level_of_education}")

lunch=df["lunch"].unique()
print(f"unique value in lunch:{lunch}")

test_preparation_course=df["test_preparation_course"].unique()
print(f"unique value in test_preparation_course:{test_preparation_course}")





```

```python
# define numerical & categorical columns
numeric_features = [feature for feature in df.columns if df[feature].dtype != 'O']
categorical_features = [feature for feature in df.columns if df[feature].dtype == 'O']

# print columns
print('We have {} numerical features : {}'.format(len(numeric_features), numeric_features))
print('\nWe have {} categorical features : {}'.format(len(categorical_features), categorical_features))
```

```python
df["Total Marks"]=df["math_score"] + df["writing_score"] + df["reading_score"]
df["Averaeg Score"]=df["Total Marks"]/3
df.head(1)
```

```python
Maths_full_score=df[df["math_score"]==100]["Averaeg Score"].count()
print(Maths_full_score)
```

```python
reading_less_20=df[df["reading_score"]<20]["Averaeg Score"].count()
print(reading_less_20)
```
